SQL queries use easy to understand key words like select, from, limit, where, and, or, as, and more.

# SQL Key Concepts

## 2. SQL is not case sensitive like Python.

SQL is not case sensitive like Python. This means that key words, table names, and column names can include a mixture of lower-case and upper-case letters without changing the meaning of those words. However, by convention, SQL commands are capitalized to signify they are not names from tables in the database.

# SQL Key Concepts

## 3. Special characters, like asterisks and percentage signs, are used to identify specific patterns.

SQL makes use of special characters, like asterisks and percentage signs, to identify specific patterns.

# SQL Key Concepts

## 4. A SQL query is often divided onto multiple lines to make it easier to read.

A SQL query is often divided onto multiple lines to make it easier to read. This is not required, but is a useful convention.

# SQL Key Concepts

## 5. A SQL query ends with a semicolon.

A SQL query ends with a semicolon. This is useful so that you can use multiple lines for longer queries.

## 6. Pay attention to details.

Pay attention to details. Little things, like a missing space or a missing comma, or the order of the commands can prevent the query from running properly, so pay attention to those syntactical details.
