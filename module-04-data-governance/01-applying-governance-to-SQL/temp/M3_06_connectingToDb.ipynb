# Connecting and Disconnecting to a Database

## 1. Connecting and Disconnecting to a SQLite Database

```python
import pandas as pd
import sqlite3 as sql
con = sql.connect('teca.db')
```

```python
pd.read_sql('SELECT * FROM parents LIMIT 5;', con)
```

```python
pd.read_sql('SELECT * FROM sites LIMIT 5;', con)
```

```python
con.close()
```

```python
pd.read_sql('SELECT * FROM categories LIMIT 5;', con)
```

## 2. Connecting and Disconnecting with a Context Manager 

When you're not exploring the data, then this approach is best. 

```python
with sql.connect('teca.db') as con:
    df = pd.read_sql('SELECT * FROM categories LIMIT 5;', con)
df
```
