# Descriptive SQL Queries in Python

## Import Modules and Connect to Database

```python
import pandas as pd
import sqlite3 as sql
```

```python
con = sql.connect('teca.db')
```

## Get the Database Schema

```python
df = pd.read_sql('SELECT * FROM sqlite_master WHERE type="table";', con)
df = df.set_index('name')
df
```

```python
print(df.loc['transactions', 'sql'])
```

```python
pd.read_sql('SELECT * FROM transactions LIMIT 5;', con)
```

## Numeric Columns

```python
pd.read_sql('SELECT SUM(units) AS total_units FROM transactions;', con)
```

```python
query = '''
SELECT SUM(units) AS total_units, 
MIN(units) AS min_units,
AVG(units) as mean_units,
MAX(units) as max_units
FROM transactions;
'''
pd.read_sql(query, con)
```

```python
query = '''
SELECT  
MIN(unformatted_date), 
MAX(unformatted_date) 
FROM transactions;
'''
pd.read_sql(query, con)
```

## Categorical Columns

```python
query = '''
SELECT 
DISTINCT(product_id)
FROM transactions;
'''
pd.read_sql(query, con)
```

```python
query = '''
SELECT 
COUNT(DISTINCT(product_id))
FROM transactions;
'''
pd.read_sql(query, con)
```
