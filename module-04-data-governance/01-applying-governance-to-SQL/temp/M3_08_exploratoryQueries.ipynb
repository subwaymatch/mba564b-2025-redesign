# Exploratory Database Queries in Python

## Import Modules and Connect to Database

```python
import pandas as pd
import sqlite3 as sql
```

```python
con = sql.connect('teca.db')
```

## Get the Database Schema

```python
df = pd.read_sql('SELECT * FROM sqlite_master WHERE type="table";', con)
df = df.set_index('name')
df
```

## Table Overview

### Column Names, Types, and Relationships

```python
print(df.loc['transactions', 'sql'])
```

### Number of Rows

```python
pd.read_sql('SELECT COUNT(*) FROM transactions;', con)
```

### First Five Rows

```python
pd.read_sql('SELECT * FROM transactions LIMIT 5;', con)
```

### Number of Missing Values

```python
pd.read_sql('SELECT COUNT(*) FROM transactions WHERE customer_id IS "nan";', con)
```

## Read in the Data for Analysis

```python
query = '''
SELECT *
FROM transactions
WHERE revenue > 2;
'''
df = pd.read_sql(query, con)
df.info()
```
