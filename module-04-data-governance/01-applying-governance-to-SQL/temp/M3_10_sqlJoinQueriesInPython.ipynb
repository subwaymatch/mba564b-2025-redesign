# SQL Join Queries in Python

## Import Modules and Connect to Database

```python
import pandas as pd
import sqlite3 as sql
```

```python
con = sql.connect('teca.db')
```

## Get the Database Schema

```python
df = pd.read_sql('SELECT * FROM sqlite_master WHERE type="table";', con)
df = df.set_index('name')
df
```

```python
print(df.loc['transactions', 'sql'])
```

```python
print(df.loc['sites', 'sql'])
```

## Left Join

```python
query = '''
SELECT t.revenue, t.costs, t.units, s.site_name
FROM transactions AS t
LEFT JOIN sites as s
ON t.site_id=s.site_id;
'''
df = pd.read_sql(query, con)
df
```

```python
query = '''
SELECT t.revenue, t.costs, t.units, s.site_name
FROM transactions AS t
LEFT JOIN sites as s
ON t.site_id=s.site_id 
WHERE site_name = "433 Bay Minette" 
ORDER BY revenue DESC;
'''
df = pd.read_sql(query, con)
df
```

```python
query = '''
SELECT t.revenue, t.costs, t.units, s.site_name, p.product_name
FROM transactions AS t
LEFT JOIN sites as s
ON t.site_id=s.site_id
LEFT JOIN products as p
ON t.product_id=p.product_id;
'''
df = pd.read_sql(query, con)
df
```
